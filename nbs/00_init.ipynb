# Package init

> Number→glyph web UI and Bluetooth receive for the iverson.io app.

Lazy re-exports of the BLE constants and data paths used by iverson.io + `iverson-receive`.


```python
#| default_exp __init__
```

```python
#| export
__version__ = "0.1.0"

__all__ = ["SERVICE", "CHAR", "NAME", "data_dir", "asset_dir"]

_lazy = {
    "SERVICE": ("bt", "SERVICE"),
    "CHAR": ("bt", "CHAR"),
    "NAME": ("bt", "NAME"),
    "data_dir": ("core", "data_dir"),
    "asset_dir": ("core", "asset_dir"),
}

def __getattr__(name):
    if name not in _lazy:
        raise AttributeError(f"module {__name__!r} has no attribute {name!r}")
    import importlib
    module, attr = _lazy[name]
    value = getattr(importlib.import_module(f".{module}", __name__), attr)
    globals()[name] = value
    return value

def __dir__():
    return sorted(set(globals()) | set(_lazy))
```
